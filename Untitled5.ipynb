```python
from pathlib import Path
import pandas as pd
from IPython.display import display
```

```python
tracks = [
    "Web Development",
    "Data Science",
    "Artificial Intelligence",
    "Cybersecurity",
    "UI/UX Design",
]

universities = [
    "CASE",
    "NUST",
    "COMSATS",
    "FAST",
    "Punjab University",
]

rows = []

for i in range(80):
    rows.append({
        "Applicant Name": f"Demo Applicant {i + 1:03d}",
        "Email": f"applicant{i + 1:03d}@example.com",
        "Phone": f"0300{i + 1:07d}",
        "Domain Applied": tracks[i % 5],
        "University": universities[i % 5],
        "Application Date": (
            pd.Timestamp("2026-05-01")
            + pd.Timedelta(days=i % 28)
        ).strftime("%Y-%m-%d"),
        "Status": ["Under Review", "Selected", "Rejected"][i % 3],
    })

mock = pd.DataFrame(rows)


mock.loc[0, "Applicant Name"] = None
mock.loc[1, "Email"] = None
mock.loc[2, "University"] = None
mock.loc[3, "Status"] = None
mock.loc[4, "Domain Applied"] = " WEB DEV "
mock.loc[5, "Domain Applied"] = "data sci"
mock.loc[6, "Phone"] = "0300-000 0007"
mock.loc[7, "Application Date"] = "12/05/2026"
mock.loc[8, "Application Date"] = "invalid-date"
mock.loc[9, "Applicant Name"] = "  Demo Applicant 010  "
mock.loc[10, "Email"] = " APPLICANT011@EXAMPLE.COM "


mock.loc[11, "Email"] = " APPLICANT013@EXAMPLE.COM "

mock = pd.concat(
    [mock, mock.iloc[[20, 21]]],
    ignore_index=True,
)


if Path("applicants.csv").exists():
    raise FileExistsError(
        "applicants.csv already exists. Use that file or rename it first."
    )

mock.to_csv("applicants.csv", index=False)

print("Created fictional practice dataset:", mock.shape)
```

```python
raw = pd.read_csv("applicants.csv", dtype="string")
df = raw.copy()

print("Original shape:", raw.shape)
display(raw.head())

raw.info()

print("\nMissing values:")
display(raw.isnull().sum())

print("Exact duplicate rows:", raw.duplicated().sum())

print("\nOriginal domain values:")
display(raw["Domain Applied"].value_counts(dropna=False))
```

```python
starting_rows = len(df)
before_missing = raw.isnull().sum()

exact_duplicates_removed = int(df.duplicated().sum())
df = df.drop_duplicates().copy()

for column in df.columns:
    df[column] = df[column].str.strip()

df = df.replace(r"^\s*$", pd.NA, regex=True)

# Normalize email casing before checking repeated applicants.
df["Email"] = df["Email"].str.lower()

print("Exact duplicates removed:", exact_duplicates_removed)
display(df.isnull().sum())
```

```python
missing_name_rows = int(df["Applicant Name"].isna().sum())
df = df.dropna(subset=["Applicant Name"]).copy()

missing_email_rows = int(df["Email"].isna().sum())
df = df.dropna(subset=["Email"]).copy()

university_filled = int(df["University"].isna().sum())
status_filled = int(df["Status"].isna().sum())

df["University"] = df["University"].fillna("Not Specified")
df["Status"] = df["Status"].fillna("Unknown")
```

```python
email_duplicates_removed = int(
    df.duplicated(subset=["Email"]).sum()
)

df = df.drop_duplicates(
    subset=["Email"],
    keep="first",
).copy()

print("Additional email duplicates removed:", email_duplicates_removed)
```

```python
domain_mapping = {
    "web dev": "Web Development",
    "webdev": "Web Development",
    "web development": "Web Development",
    "data sci": "Data Science",
    "data science": "Data Science",
    "ai": "Artificial Intelligence",
    "artificial intelligence": "Artificial Intelligence",
    "cyber security": "Cybersecurity",
    "cybersecurity": "Cybersecurity",
    "ui ux": "UI/UX Design",
    "ui/ux": "UI/UX Design",
    "ui/ux design": "UI/UX Design",
}

domain_keys = df["Domain Applied"].str.lower()
mapped_domains = domain_keys.map(domain_mapping)

unmapped_domains = (
    df["Domain Applied"].notna() & mapped_domains.isna()
)

if unmapped_domains.any():
    raise ValueError(
        "Extend domain_mapping for: "
        + str(df.loc[unmapped_domains, "Domain Applied"].unique().tolist())
    )

df["Domain Applied"] = mapped_domains.astype("string")

status_mapping = {
    "under review": "Under Review",
    "selected": "Selected",
    "rejected": "Rejected",
    "unknown": "Unknown",
}

status_keys = df["Status"].str.lower()
mapped_statuses = status_keys.map(status_mapping)

if mapped_statuses.isna().any():
    raise ValueError(
        "Extend status_mapping for: "
        + str(
            df.loc[mapped_statuses.isna(), "Status"].unique().tolist()
        )
    )

df["Status"] = mapped_statuses.astype("string")

university_mapping = {
    "case": "Sir Syed CASE Institute of Technology",
    "ss case it": "Sir Syed CASE Institute of Technology",
    "nust": "NUST",
    "comsats": "COMSATS",
    "fast": "FAST",
    "punjab university": "Punjab University",
}

university_keys = df["University"].str.lower()
df["University"] = (
    university_keys.map(university_mapping)
    .fillna(df["University"])
    .astype("string")
)

print("Clean domain values:")
display(df["Domain Applied"].unique())
```

```python
original_dates = df["Application Date"].copy()

parsed_dates = pd.Series(
    pd.NaT,
    index=df.index,
    dtype="datetime64[ns]",
)

for date_format in ["%Y-%m-%d", "%d/%m/%Y", "%d-%m-%Y"]:
    candidate = pd.to_datetime(
        original_dates,
        format=date_format,
        errors="coerce",
    )
    parsed_dates = parsed_dates.fillna(candidate)

invalid_dates = original_dates.notna() & parsed_dates.isna()
invalid_date_count = int(invalid_dates.sum())

print("Unparseable original dates:")
display(df.loc[invalid_dates, ["Email", "Application Date"]])

df["Application Date"] = parsed_dates

# Remove common formatting characters without converting to numbers.
df["Phone"] = df["Phone"].str.replace(
    r"[\s()+-]",
    "",
    regex=True,
)

df["Phone"] = df["Phone"].replace("", pd.NA)

df.info()
```

```python
print("Applications by domain:")
display(df["Domain Applied"].value_counts(dropna=False))

print("Applications by status:")
display(df["Status"].value_counts(dropna=False))

print("Applications by university:")
display(df["University"].value_counts(dropna=False))
```

```python
assert df["Applicant Name"].notna().all()
assert df["Email"].notna().all()
assert not df.duplicated().any()
assert not df["Email"].duplicated().any()
assert pd.api.types.is_datetime64_any_dtype(df["Application Date"])

expected_domains = {
    "Web Development",
    "Data Science",
    "Artificial Intelligence",
    "Cybersecurity",
    "UI/UX Design",
}

actual_domains = set(df["Domain Applied"].dropna())

assert actual_domains == expected_domains, (
    f"Review domain coverage: {actual_domains}"
)

print("Final shape:", df.shape)
print("\nRemaining missing values:")
display(df.isnull().sum())

print("All validation checks passed.")
```

```python
print(f"Starting dataset: {starting_rows} rows, {raw.shape[1]} columns")
print(f"Exact duplicates removed: {exact_duplicates_removed}")
print(f"Rows dropped for missing name: {missing_name_rows}")
print(f"Rows dropped for missing email: {missing_email_rows}")
print(f"Additional email duplicates removed: {email_duplicates_removed}")
print(f"Missing universities filled: {university_filled}")
print(f"Missing statuses filled: {status_filled}")
print(f"Unparseable dates retained as NaT: {invalid_date_count}")
print(f"Final dataset: {len(df)} rows, {df.shape[1]} columns")
```

```python
df.to_csv(
    "applicants_cleaned.csv",
    index=False,
    date_format="%Y-%m-%d",
)

print("Saved applicants_cleaned.csv")
```
